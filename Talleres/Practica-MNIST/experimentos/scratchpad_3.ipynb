#Red Densa

```
import numpy as np
import tensorflow as tf
(x_train, y_train), (x_test, y_test)=tf.keras.datasets.mnist.load_data()
x_train=x_train.reshape(x_train.shape[0], 28*28)
x_test=x_test.reshape(x_test.shape[0], 28*28)
import matplotlib.pyplot as plt
plt.imshow(x_train[0].reshape(28,28), cmap='gray')
```

```
In=tf.keras.layers.Input(shape=(784,))
Cp1=tf.keras.layers.Dense(64,activation='relu')(In)
Cp2=tf.keras.layers.Dense(64,activation='relu')(Cp1)
Out=tf.keras.layers.Dense(10,activation='softmax')(Cp2)
Model=tf.keras.Model(inputs=In,outputs=Out)
```

```
Model.summary()
```

```
Model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
Model.fit(x_train,y_train,epochs=30,batch_size=32)
```

#Convolucional

```
import numpy as np
import tensorflow as tf
(x_train, y_train), (x_test, y_test)=tf.keras.datasets.mnist.load_data()

x_train = x_train[..., np.newaxis]

print("Shape de x_train:", x_train.shape)
```

```
In=tf.keras.layers.Input(shape=(28,28,1),name="Input")
conv1=tf.keras.layers.Conv2D(32,kernel_size=3,activation='relu',strides=1,padding="same",name="Convolucion_1")(In)
max=tf.keras.layers.MaxPooling2D(pool_size=(2,2),name="Pooling_1")(conv1)
conv2=tf.keras.layers.Conv2D(64,kernel_size=3,activation='relu',strides=1,padding="same",name="Convolucion_2")(max)
max2=tf.keras.layers.MaxPooling2D(pool_size=(2,2),name="Pooling_2")(conv2)
f=tf.keras.layers.Flatten(name="Flatten")(max2)
Out=tf.keras.layers.Dense(10,activation='softmax',name="Densa")(f)
Model=tf.keras.Model(inputs=In,outputs=Out)
```

```
Model.summary()
```

```
Model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
Model.fit(x_train,y_train,epochs=30,batch_size=32)
```

```
WC1=Model.get_layer("Convolucion_1").get_weights()[0]
import matplotlib.pyplot as plt
plt.imshow(WC1[:,:,0,2],cmap='gray')
```

```
model_conv_1=tf.keras.Model(inputs=Model.input,outputs=Model.layers[3].output)
```

```
y_conv1=model_conv_1.predict(x_train[0].reshape(1,28,28,1))
plt.imshow(y_conv1[0,:,:,0],cmap='gray')
#
```
